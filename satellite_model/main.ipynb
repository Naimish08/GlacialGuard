```python
import cv2
import matplotlib.pyplot as plt
import os
```

```python
# Define the dataset path
dataset_path = '../glacial-lake-dataset/images/'

def read_image(image_name):
    """
    Read image using OpenCV and convert from BGR to RGB
    """
    img_path = os.path.join(dataset_path, image_name)
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB
    return img

```

```python
# Example: Read and display an image
image_name = 'image_1.png'  # Replace with your image name
img = read_image(image_name)
```
